#### 5.- De la lista de accessions que se proporciona obtén lo siguiente:
1. Los que tengan el número 5
2. Los que tengan la letra 'd' o la 'e'
3. Los que tengan las letras 'd' y 'e' en ese orden
4. Los que tengan las letras 'd' y 'e' en ese orden con una letra cualquiera en medio
5. Los que tengan ambas letras 'd' y 'e' en cualquier orden
6. Los que empiezan por 'x' o por 'y'
7. Los que empiezan por 'x' o 'y' y terminan por 'e'
8. Los que tienen tres o más dígitos seguidos
9. Los que terminan con 'd' seguidos de 'a' o 'r' o 'p'

```python
import re
```

```python
accs = ['xkn59438', 'yhdck2', 'eihd39d9', 'chdsye847', 'hedle3455', 'xjhd53e', '45da', 'de37dp']
```

```python
#1.
for elemento in accs:
    if re.search(r"5",elemento):
        print(elemento)
```

```python
#2. OPCION 1
for elemento in accs:
    if re.search(r"d",elemento) or re.search(r"e",elemento):
        print(elemento)
```

```python
#2. OPCION 2
for elemento in accs:
    if re.search(r"d|e",elemento):
        print(elemento)
```

```python
#2. OPCION 3
for elemento in accs:
    if re.search(r"[de]",elemento):
        print(elemento)
```

```python
#3
for elemento in accs:
    if re.search(r"de",elemento):
        print(elemento)
```

```python
#4
for elemento in accs:
    if re.search(r"d.*e",elemento):
        print(elemento)
```

```python
# *puede no estar o estar infinito numero de veces, + tiene q estar por lo menos una vez 
for elemento in accs:
    if re.search(r"d.+e",elemento):
        print(elemento)
```

```python
#5 OPCIÓN 1
for elemento in accs:
    if re.search(r"[de].*[de]",elemento):
        print(elemento)
```

```python
#5 OPCION 2
for elemento in accs:
    if re.search(r"d",elemento) and re.search(r"e",elemento):
        print(elemento)
```

```python
#6
for elemento in accs:
    if re.search(r"^x|^y",elemento):
        print(elemento)
#también con el or 
```

```python
#7 OPCIÓN 1
for elemento in accs:
    if re.search(r"^[xy]",elemento) and re.search(r"e$", elemento):
        print(elemento)
```

```python
#7 OPCIÓN 2
for elemento in accs:
    if re.search(r"^[xy].*e$",elemento):
        print(elemento)
```

```python
#8 OPCIÓN 1
for elemento in accs:
    if re.search(r"\d{3}",elemento):
        print(elemento)
#{3} es de 3 a infinito, si es {3,} es hasta 3 
```

```python
#8 OPCIÓN 2
for elemento in accs:
    if re.search(r"[0-9]{3}",elemento):
        print(elemento)
```

```python
#9
for elemento in accs:
    if re.search(r"d[arp]$",elemento):
        print(elemento)
```
